## Evaluating Lush & Glow AI-Driven Smart Scheduling

```python
import pandas as pd
import numpy as np
import scipy.stats as stats
import statsmodels.api as sm
import statsmodels.formula.api as smf

# Load Datasets
appointments = pd.read_csv('appointment_records.csv')
df_exp = pd.read_csv('experiment_dataset.csv')
df_retro = pd.read_csv('retrospective_dataset.csv')

# 1. Guardrail Analysis
print(f"No-Show Rate (Control): {df_exp[df_exp['group']=='control']['no_show_count'].mean():.4f}")
print(f"No-Show Rate (Treatment): {df_exp[df_exp['group']=='treatment']['no_show_count'].mean():.4f}")

# 2. Retrospective DiD
retro_model = smf.ols('rpv ~ C(group) * C(period)', data=df_retro).fit()
print(retro_model.summary())
```
